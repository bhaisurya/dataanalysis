```python
import pandas as pd
import zipfile

# Open the ZIP archive and extract the specific CSV file
with zipfile.ZipFile("Spotify+Streaming+History.zip") as z:
    with z.open("spotify_history.csv") as f:
        df = pd.read_csv(f)

print("Dataset shape:", df.shape)
print("DATASET INFORMATION")
df.info()

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE ROWS")
print("Number of duplicates:", df.duplicated().sum())

print("\nDATA TYPES")
print(df.dtypes)
```

```python
# Display some duplicated records, including their matching copies
duplicate_rows = df[df.duplicated(keep=False)]

display(
    duplicate_rows
    .sort_values("ts")
    .head(10)
)

print("Total rows involved in duplicate groups:", len(duplicate_rows))
```

```python
missing_reasons = df[
    df["reason_start"].isna() |
    df["reason_end"].isna()
]

display(missing_reasons.head(10))

print("Rows with missing reason values:", len(missing_reasons))
```

```python
df["reason_start"] = df["reason_start"].fillna("Unknown")
df["reason_end"] = df["reason_end"].fillna("Unknown")

print("Missing values after filling:")
print(df[["reason_start", "reason_end"]].isnull().sum())
df["reason_start"].fillna("Unknown")
print("Rows before removing duplicates:", len(df))

df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
print("Duplicates remaining:", df.duplicated().sum())
df["ts"] = pd.to_datetime(df["ts"], errors="coerce")

print("Timestamp data type:", df["ts"].dtype)
print("Invalid timestamps:", df["ts"].isna().sum())
print(df["ms_played"].describe())

print(
    "Negative playback durations:",
    (df["ms_played"] < 0).sum()
)
```

```python
# Check leading/trailing whitespace in text columns
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    count = (
        df[col].notna() &
        (df[col] != df[col].str.strip())
    ).sum()

    print(f"{col}: {count} values with extra whitespace")
    # Inspect distinct values in selected categorical columns
for col in ["platform", "reason_start", "reason_end"]:
    print(f"\n{col}:")
    print(df[col].value_counts().head(15))
    print("FINAL DATASET SHAPE:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

df.to_csv("spotify_history_cleaned.csv", index=False)
from google.colab import files
files.download("spotify_history_cleaned.csv")
```
